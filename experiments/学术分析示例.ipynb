```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import statsmodels.api as sm

# 设置中文显示
plt.rcParams['font.sans-serif'] = ['SimHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

print('✅ 学术分析环境已就绪')
print(f'pandas: {pd.__version__}')
print(f'numpy: {np.__version__}')
```

```python
# 示例：基本统计分析
# 生成示例数据
np.random.seed(42)
data = {
    'group': ['Control'] * 50 + ['Treatment'] * 50,
    'score': np.concatenate([
        np.random.normal(100, 15, 50),
        np.random.normal(110, 15, 50)
    ])
}
df = pd.DataFrame(data)

# 描述统计
print('=== 描述统计 ===')
print(df.groupby('group')['score'].describe())

# t 检验
control = df[df['group'] == 'Control']['score']
treatment = df[df['group'] == 'Treatment']['score']
t_stat, p_value = stats.ttest_ind(control, treatment)
print(f'\n=== t 检验 ===')
print(f't = {t_stat:.3f}, p = {p_value:.4f}')
```

```python
# 可视化
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# 箱线图
df.boxplot(column='score', by='group', ax=axes[0])
axes[0].set_title('Score by Group')
axes[0].set_xlabel('Group')
axes[0].set_ylabel('Score')

# 直方图
for group in ['Control', 'Treatment']:
    subset = df[df['group'] == group]['score']
    axes[1].hist(subset, alpha=0.6, label=group, bins=15)
axes[1].set_title('Score Distribution')
axes[1].set_xlabel('Score')
axes[1].legend()

plt.suptitle('Academic Data Analysis Example')
plt.tight_layout()
plt.savefig('analysis_example.png', dpi=150)
print('📊 图表已保存: analysis_example.png')
```
